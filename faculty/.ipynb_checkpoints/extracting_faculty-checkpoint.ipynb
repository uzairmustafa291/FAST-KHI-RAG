```python
import requests
import json
from bs4 import BeautifulSoup
```

```python
class Faculty:

    def __init__(self,name,designation,email,extension,profile):

        self.name=name
        self.designation=designation
        self.email=email
        self.extension=extension
        self.profile=profile

    def to_dict(self):

        return {
            "name":self.name,
            "designation":self.designation,
            "email":self.email,
            "extension":self.extension,
            "profile":self.profile
        }
```

```python
departments={

    "Computer_Science":"https://khi.nu.edu.pk/faculty-php/",

    "Cyber_Security":"https://khi.nu.edu.pk/department-of-cyber-security/",

    "Artificial_Intelligence":"https://khi.nu.edu.pk/department-of-artificial-intelligence/",

    "Software_Engineering":"https://khi.nu.edu.pk/department-of-software-engineering/",


    "Electrical_Engineering":"https://khi.nu.edu.pk/department-of-electrical-engineering/",

    "Management_Sciences":"https://khi.nu.edu.pk/department-of-management-sciences/",

    "Sciences_Humanities":"https://khi.nu.edu.pk/department-of-sciences-humanities/"
}
```

```python
def download_html(url,filename):

    response=requests.get(url)

    with open(filename,"w",encoding="utf8") as f:

        f.write(response.text)
```

```python
def parse_faculty(html_file):

    soup=BeautifulSoup(open(html_file,encoding="utf8"),"html.parser")

    faculty_list=[]

    cards=soup.find_all("div",class_="gdlr-core-personnel-list")

    for card in cards:

        name=card.find(
            "h3",
            class_="gdlr-core-personnel-list-title"
        ).get_text(strip=True)

        designation=card.find(
            "div",
            class_="gdlr-core-personnel-list-position"
        ).get_text(strip=True)

        email_div=card.find(
            "div",
            class_="kingster-type-email"
        )

        email_div.find("i").extract()

        email=email_div.get_text(strip=True)

        phone_div=card.find(
            "div",
            class_="kingster-type-phone"
        )

        phone_div.find("i").extract()

        extension=phone_div.get_text(strip=True)

        profile=card.find(
            "a",
            class_="gdlr-core-personnel-list-button"
        )["href"]

        faculty=Faculty(
            name,
            designation,
            email,
            extension,
            profile
        )

        faculty_list.append(faculty)

    return faculty_list
```

```python
def save_json(faculty_list,file):

    data=[f.to_dict() for f in faculty_list]

    with open(file,"w",encoding="utf8") as fp:

        json.dump(data,fp,indent=4)
```

```python
def save_txt(faculty_list,file):

    with open(file,"w",encoding="utf8") as f:

        for member in faculty_list:

            f.write(
f"""
Name: {member.name}
Designation: {member.designation}
Email: {member.email}
Extension: {member.extension}
Profile: {member.profile}

"""
            )
```

```python
for dept,url in departments.items():

    print(f"Downloading {dept}")

    html_file=f"all_{dept}_Faculty.html"

    json_file=f"{dept}_faculty_file.json"

    txt_file=f"{dept}_faculty_file.txt"

    download_html(url,html_file)

    faculty=parse_faculty(html_file)

    save_json(faculty,json_file)

    save_txt(faculty,txt_file)

    print(f"Finished {dept}")
```
