```python
import os
import json
import uuid
from pathlib import Path

import chromadb
from sentence_transformers import SentenceTransformer
from groq import Groq
from dotenv import load_dotenv

from tqdm import tqdm
```

```python
load_dotenv()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

print(GROQ_API_KEY[:15] + "...")
```

```python
client = Groq(
    api_key=GROQ_API_KEY
)

print("Groq Connected Successfully")
```

```python
faculty_folder = Path("faculty")

json_files = sorted(faculty_folder.glob("*.json"))
txt_files = sorted(faculty_folder.glob("*.txt"))

print(f"JSON Files : {len(json_files)}")
print(f"TXT Files  : {len(txt_files)}")
```

```python
documents = []

for file in json_files:

    with open(file, "r", encoding="utf-8") as f:

        data = json.load(f)

        documents.append(
            {
                "source": file.name,
                "text": json.dumps(data, indent=2)
            }
        )

for file in txt_files:

    with open(file, "r", encoding="utf-8") as f:

        documents.append(
            {
                "source": file.name,
                "text": f.read()
            }
        )

print("Total Documents :", len(documents))
```

```python
print(documents[0]["source"])
print("-" * 60)
print(documents[0]["text"][:1000])
```

```python
def chunk_text(text, chunk_size=500, overlap=100):
    """
    Split text into overlapping chunks.
    """

    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunk = text[start:end]

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
chunked_documents = []

for doc in documents:

    chunks = chunk_text(doc["text"])

    for i, chunk in enumerate(chunks):

        chunked_documents.append(
            {
                "id": str(uuid.uuid4()),
                "source": doc["source"],
                "chunk_id": i,
                "text": chunk
            }
        )

print("Total Chunks:", len(chunked_documents))
```

```python
print(chunked_documents[0]["source"])

print("-"*60)

print(chunked_documents[0]["text"])
```

```python
embedding_model = SentenceTransformer(
    "BAAI/bge-small-en-v1.5"
)

print("Embedding model loaded.")
```

```python
chroma_client = chromadb.PersistentClient(
    path="chroma_db"
)
```

```python
collection = chroma_client.get_or_create_collection(
    name="faculty_rag"
)

print("Collection Ready")
```

```python
for doc in tqdm(chunked_documents):

    embedding = embedding_model.encode(
        doc["text"]
    ).tolist()

    collection.add(

        ids=[doc["id"]],

        embeddings=[embedding],

        documents=[doc["text"]],

        metadatas=[
            {
                "source": doc["source"],
                "chunk_id": doc["chunk_id"]
            }
        ]
    )

print("Finished storing embeddings.")
```

```python
print(collection.count())
```

```python
def retrieve(query, top_k=5):

    query_embedding = embedding_model.encode(query).tolist()

    results = collection.query(

        query_embeddings=[query_embedding],

        n_results=top_k
    )

    return results
```

```python
results = retrieve(
    "all the phd holders?"
)
```

```python
for i in range(len(results["documents"][0])):

    print("=" * 80)

    print("Source :", results["metadatas"][0][i]["source"])

    print()

    print(results["documents"][0][i][:600])

    print()
```

```python
def build_context(results):
    """
    Combine retrieved chunks into one context string.
    """
    context = ""

    for i, doc in enumerate(results["documents"][0]):
        source = results["metadatas"][0][i]["source"]

        context += f"\n\nSource: {source}\n"
        context += doc

    return context
```

```python
def create_prompt(question, context):
    return f"""
You are a helpful university faculty assistant.

Answer ONLY using the context provided below.

If the answer is not present in the context, reply exactly:

"I don't have enough information to answer that."

Do not make up names or facts.

==========================
Context:
{context}
==========================

Question:
{question}

Answer:
"""
```

```python
def ask_groq(prompt):

    response = client.chat.completions.create(

        model="llama-3.3-70b-versatile",

        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],

        temperature=0.2
    )

    return response.choices[0].message.content
```
